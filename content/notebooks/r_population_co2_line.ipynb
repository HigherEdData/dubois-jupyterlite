### 5. Creating a Time Series Line Chart

After successfully displaying the data above, you should be able to see that it has data in 23 rows and 4 columns. Each row is an observation. Each column is a **variable**:
* **Year** is the census year of a given observation from 1800 to 2020.
* **White** containts total US white population in millions for the given year of an observation.
* **Black** containts total US Black population in millions for the given year of an observation.
* **CO2** containts annual US carbon dioxide emissions in million metric tons for the given year of an observation. We'll use this variable in step 8.

As a first step, we will create a **time series line graph** of the population data using the shortest code possible. We often want to use the simplest code possible while first exploring the data because it helps us easily adjust the lines and plots at the core of the graph. Then we add things like labels and titles later which can require a lot of distracting code.

Our short initial code will:
1. Repeat the Layer 1 code in line 3 that we wrote above to **read in** the population data, and load the **ggplot2** package of functions into our Library to generate our graph in line 4: ```library(ggplot2)```<br><br>
2. Add a new Layer 2 with a **ggplot function** in line 8 followed by open parentheses to tell R that we will plot data from the df data frame with an "aesthetic mapping" **aes** specification that maps the **Year** variable on the **x axis**: ```ggplot(df, aes(x = Year))```<br><br>
3. After the close parentheses, we add a ```+``` and then a new line of code in line 9 with ```geom_line()``` that tells ggplot we want to draw a line. Within the geom_line parentheses, we add another **aes** specification that maps the **White** variable on the **y axis** and gives the line the color label "White": ```geom_line(aes(y = White, color = "White"))```<br><br>
4. We add another ```+``` and a second ```geom_line()``` in line 10 that draws the Black population line the same way: ```geom_line(aes(y = Black, color = "Black"))```. Because each line has its own color label, ggplot draws each line in a different color and adds a legend to the right of the graph.

### 6. Creating a Similar Grid Layer to Du Bois'

When writing code for data analysis, you want to start with the minimum code necessary. Then you add more lines of code to add layers to your analysis. You can see this method in action with data visualization, where additional lines of code add visual layers to your figure.

This approach of starting simple and layering will produce code that is as simple and legible as possible for yourself and others, also producing clearer graphs.

Now that we've produced our line graph with the simplest code possible in Layers 1 and 2, we can start to add more visual layers by writing further lines of code. As we add layers, we will see why code is better for creating a record for how each visual element is added. It's harder to keep such a record with AI assistants or point and click software like Excel and Google sheets.

When plotting points (i.e. a scatter plot) or lines in 2 dimensional space, it is often helpful to add **x axis** and **y axis** grid lines. This helps the viewer to more easily see the X and Y positions for given data points or points on a line.

ggplot already draws light grid lines on a grey background by default, but only at a few widely spaced tick marks. To add more grid lines, we add a new 3rd layer that sets the tick marks, which R calls **breaks**, on each axis:
* ```scale_x_continuous()``` sets the breaks on the x axis and ```scale_y_continuous()``` sets the breaks on the y axis.
* The ```seq()``` function creates a sequence of numbers from a starting value, to an ending value, by a step size. For example, ```seq(1800, 2020, by = 20)``` creates the sequence 1800, 1820, 1840, ... 2020.
* Note the ```+``` sign at the end of line 9. It tells R that we are adding another layer to the graph.

````r
##### CODE FROM LAST STEP ####
# Layer 1: Load Data and the ggplot2 Package
df <- read.csv("d_population_co2.csv")
library(ggplot2)

# Layer 2: Create Figure for Basic Line Chart
ggplot(df, aes(x = Year)) +
    geom_line(aes(y = White, color = "White")) + # DRAW WHITE LINE
    geom_line(aes(y = Black, color = "Black")) + # DRAW BLACK LINE

##### NEW CODE BELOW ####
# New Layer 3: Add grid lines by setting the tick marks (breaks) on each axis
scale_x_continuous(breaks = seq(1800, 2020, by = 20)) + # a grid line every 20 years
scale_y_continuous(breaks = seq(0, 240, by = ____))     # fill in the blank for a grid line every 20 million people
````

<details> <summary><strong>Details on how the new code adds additional layers and chart elements.</strong></summary>

Key new lines of code do the following:

1. Set the figure's width to 11 by height to 14 with ```options(repr.plot.width=22/2, repr.plot.height=28/2)```. Adding ```jupyter.plot_scale=3``` shrinks how large the figure is displayed so that it fits on the page without scrolling. Try a larger number like 4 to make it even smaller, or 2 to make it larger.

2. Source the Du Bois theme from a file at the STEM + Du Bois site: ```source("https://stemdubois.org/theme_dubois.R")```. You can see all of the theme options in the "theme_dubois.R" theme file here: https://github.com/HigherEdData/Du-Bois-STEM/blob/main/theme_dubois.R

3. Add ```linewidth = 4``` and ```lineend = "round"``` to each ```geom_line()``` to draw thick lines with rounded ends like Du Bois' lines. The ```color =``` label in each ```geom_line()``` is the text that appears in the legend.

4. Add ```limits = c(0, 240)``` and ```expand = c(0, 0)``` to ```scale_y_continuous()``` so the y axis runs from exactly 0 to 240 million.

5. Add the ```scale_color_manual``` function of ggplot to set the color of each line based on its legend label.

6. Specify ```theme_dubois()``` on a new line after a ```+``` sign. This will apply several Du Bois style options like an "antique white" background color and minimal text in labeling axes. The Du Bois theme hides the legend, grid lines, and x-axis labels, so the ```theme()``` function after it turns them back on and uses ```legend.position = "inside"``` to place the legend inside the plot area.

7. Use the ```labs``` function to add the title and subtitle, remove the x-axis title with ```x = NULL```, and label the y axis ```"MILLIONS"```.

</details>

````r
##### CODE FROM LAST STEP ####
# Layer 1: Load Data, Library, and Source Theme
df <- read.csv("d_population_co2.csv")
library(ggplot2)
options(repr.plot.width=22/2, repr.plot.height=28/2, jupyter.plot_scale=3) # set the figure width to 11 and height to 14, shown smaller so it fits on the page
source("https://stemdubois.org/theme_dubois.R")

#### ADDITIONAL FORMATTING FOR YOU TO EDIT BELOW ######
# Layer 2: Plot the lines, now with Du Bois' line widths and a label for each line in the legend
ggplot(df, aes(x = Year)) +
    geom_line(aes(y = White, color = "WHITE - BLANCHE"), linewidth = 4, lineend = "round") +
    geom_line(aes(y = Black, color = "BLACK - NOIRE"), linewidth = 4, lineend = "round") +

# Layer 3: Grid lines every 20 years and every 20 million people
scale_x_continuous(breaks = seq(1800, 2020, by = 20)) +
scale_y_continuous(breaks = seq(0, 240, by = 20), limits = c(0, 240), expand = c(0, 0)) +

# New Layer 4: Set the color of each line in the legend
scale_color_manual(values = c("WHITE - BLANCHE" = "green", "BLACK - NOIRE" = "red")) + ### CHANGE GREEN AND RED TO ACCESSIBLE COLORS

# New Layer 5: Apply the Du Bois theme, then turn the legend, grid lines, and axis labels back on
theme_dubois() +
theme(
    plot.background = element_rect(fill = "antiquewhite2"), ### CHANGE TO PREFERRED BACKGROUND COLOR
    legend.position = "inside",            # put the legend inside the plot area
    legend.position.inside = c(0.2, 0.9),  # 20% of the way across and 90% of the way up the plot area
    legend.title = element_blank(),        # remove the legend title
    legend.background = element_blank(),   # remove the legend's white background
    legend.key = element_blank(),          # remove the grey boxes behind the legend lines
    legend.text = element_text(size = 12),
    panel.grid.major = element_line(color = "gray"), # add the grid lines back
    axis.text.x = element_text(size = 12),           # add the year labels back
    axis.text.y = element_text(size = 14),
    axis.title.y = element_text(angle = 0, vjust = 1) # put the y-axis title at the top of the axis
) +
#### ADDITIONAL FORMATTING FOR YOU TO EDIT ABOVE ######

# New Layer 6: Title and subtitle with your name
labs(
    title = "Comparative rate of increase of the White and Negro elements of the\npopulation of the United States.\n",
    subtitle = "Accroissement proportionnel des éléments blancs et noirs aux États Unis.\n
Done by Atlanta University. Extended to 2020 with U.S. Census data.\n
Recreated by _____________\n\n", ###### FILL IN YOUR NAME HERE
    x = NULL,        # no x-axis title
    y = "MILLIONS"   # y-axis title
)
````
